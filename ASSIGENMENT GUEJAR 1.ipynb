```python
df.memory_usage(deep=True)
```

```python
mem_before = df.memory_usage(deep=True).sum() / (1024 * 1024)

print("Memory Before:", round(mem_before, 4), "MB")
```

```python
df_small = df.copy()

print(df is df_small)

print("Original:", df.shape)
print("Copy:", df_small.shape)
```

```python
df_small["type"] = df_small["type"].astype("category")

df_small["rating"] = df_small["rating"].astype("category")

print("Type:", df_small["type"].dtype)
print("Rating:", df_small["rating"].dtype)

print("Original:", df["type"].dtype)
print("Optimized:", df_small["type"].dtype)
```

```python
mem_after = df_small.memory_usage(deep=True).sum() / (1024 * 1024)

print("Memory After:", round(mem_after, 4), "MB")
```

```python
print("Memory Before:", round(mem_before, 4), "MB")
print("Memory After:", round(mem_after, 4), "MB")
```

```python
pct_saved = ((mem_before - mem_after) / mem_before) * 100
print("Percentage Saved:", round(pct_saved, 2), "%")
```

```python
# 1. Pehle check function define hoga
def check(message, condition):
    assert condition(), f"❌ Check Failed: {message}"
    print(f"✅ Passed: {message}")

# 2. Phir saare self-checks run honge
check("type and rating are category in df_small", lambda: str(df_small["type"].dtype) == "category" and str(df_small["rating"].dtype) == "category")
check("raw df was not modified (df_small is a separate copy)", lambda: df_small is not df and str(df["type"].dtype) != "category")
check("mem_after < mem_before", lambda: mem_after < mem_before)

print("PART A successfully completed!")
```
